## Final_grade

The numbers assigned to each grade (A=1 through F=5) were only used to establish rank order for mapping purposes — they don't represent equal, real differences in student performance. Because of this, the mean of 3.07 is mathematically correct but doesn't have a clear, trustworthy real-world meaning; reporting it to a stakeholder would require noting this limitation.


```python
enrollment_year_mean = skyline["enrollment_year"].mean()
print(f"Mean enrollment year: {enrollment_year_mean:.2f}")
```

```python
enrollment_year_median = skyline["enrollment_year"].median()
print(f"Median enrollment year: {enrollment_year_median:.2f}")
```

```python
enrollment_year_mode = skyline["enrollment_year"].mode()[0]
print(f"Mode enrollment year: {enrollment_year_mode}")
```

```python
enrollment_year_std = skyline["enrollment_year"].std()
print(f"Standard deviation of enrollment years: {enrollment_year_std:.2f}")
```

```python
enrollment_year_q1 = skyline["enrollment_year"].quantile(0.25)
enrollment_year_q3 = skyline["enrollment_year"].quantile(0.75)
enrollment_year_iqr = enrollment_year_q3 - enrollment_year_q1

print(f"Q1 (25th percentile): {enrollment_year_q1:.2f}")
print(f"Q3 (75th percentile): {enrollment_year_q3:.2f}")
print(f"IQR: {enrollment_year_iqr:.2f}")
```

```python
Hours_studied_mean = skyline["Hours_studied"].mean()
print(f"Mean Hours studied: {Hours_studied_mean:.2f}")
```

```python
Hours_studied_median = skyline["Hours_studied"].median()
print(f"Median Hours studied: {Hours_studied_median:.2f}")
```

```python
Hours_studied_mode = skyline["Hours_studied"].mode()[0]
print(f"Mode Hours studied: {Hours_studied_mode}")
```

```python
Hours_studied_std = skyline["Hours_studied"].std()
print(f"Standard deviation of Hours studied: {Hours_studied_std:.2f}")
```

```python
Hours_studied_q1 = skyline["Hours_studied"].quantile(0.25)
Hours_studied_q3 = skyline["Hours_studied"].quantile(0.75)
Hours_studied_iqr = Hours_studied_q3 - Hours_studied_q1

print(f"Q1 (25th percentile): {Hours_studied_q1:.2f}")
print(f"Q3 (75th percentile): {Hours_studied_q3:.2f}")
print(f"IQR: {Hours_studied_iqr:.2f}")
```

## Hours_studied

The mean (30.18) and median (30.06) are very close, suggesting the distribution is symmetrical around the center, with no significant skew. This makes sense since Hours_studied was generated from a normal distribution.
